```python
#Import das Bibliotecas
import numpy as np
import binascii
import re
```

```python
#Definição da Função de Ativação de Hopfield (Bipolar)
def activation_function(x):
    if x < 0:
        return -1
    return 1
```

```python
#Em seguida, Cria-se a Matrix de recuperação de informação
class Matrix:

    @staticmethod
    def matrix_vector_multiplication(matrix, vector):
        return matrix.dot(vector)

    @staticmethod
    def clear_diagonal(matrix):
        np.fill_diagonal(matrix, 0)
        return matrix

    @staticmethod
    def outer_product(pattern):
        return np.outer(pattern, pattern)

    @staticmethod
    def add_matrices(matrix1, matrix2):
        return matrix1 + matrix2
```

```python
#Em seguida, defini-se a estrutura da Rede de Hopfield, juntamente com seus métodos de Treinamento e Recuperação da Informação
class HopfieldNetwork:

    def __init__(self, dimension):
        self.weight_matrix = np.zeros((dimension, dimension))

    def train(self, pattern):

        pattern_bipolar = HopfieldNetwork.transform(pattern)

        pattern_weight_matrix = Matrix.outer_product(pattern_bipolar)

        pattern_weight_matrix = Matrix.clear_diagonal(pattern_weight_matrix)

        self.weight_matrix = Matrix.add_matrices(self.weight_matrix, pattern_weight_matrix)

    def recall(self, pattern):

        pattern_bipolar = HopfieldNetwork.transform(pattern)

        result = Matrix.matrix_vector_multiplication(self.weight_matrix, pattern_bipolar)

        result = np.array([activation_function(x) for x in result])

        result = HopfieldNetwork.re_transform(result)

        return(result)

    @staticmethod
    def transform(pattern):
        return np.where(pattern == 0, -1, pattern)

    @staticmethod
    def re_transform(pattern):
        return np.where(pattern == -1, 0, pattern)
```

<b> Exemplo de Aplicação com Hopfield </b>
    
Valores para Teste:

- Array de Entrada = [0,0,0,0,0,0,1,1] (Treinamento)
- Array com Ruído = [0,1,1,0,0,0,0,0]

```python
#Inicialmente, define-se quantos elementos serão fornecidos como Entrada para estabelecer um padrão de recuperação
#Neste Exemplo, serão 8 elementos
network = HopfieldNetwork(8)
```

```python
#Realiza-se o treinamento passando o padrão de entrada.
std = np.array([0,0,0,0,0,0,1,1])
```

```python
#Treinamento
network.train(std)
```

```python
#Desta forma, ao enviar um padrão com ruído, a rede irá recuperar a informação anterior
noise = np.array([0,1,1,0,0,0,0,1])
print(noise)
```

```python
#Recuperação da Informação
network.recall(noise)
```

<b> Construção das Funções que serão utilizadas no Experimento

```python
#Funções para Binarizar os Caracteres
def text_to_bits(text, encoding='utf-8', errors='surrogatepass'):
    bits = bin(int(binascii.hexlify(text.encode(encoding, errors)), 16))[2:]
    return bits.zfill(8 * ((len(bits) + 7) // 8))

def text_from_bits(bits, encoding='utf-8', errors='surrogatepass'):
    n = int(bits, 2)
    return int2bytes(n).decode(encoding, errors)

def int2bytes(i):
    hex_string = '%x' % i
    n = len(hex_string)
    return binascii.unhexlify(hex_string.zfill(n + (n & 1)))
```

```python
#Criando uma Função para Converter a String do Hash em Binário e em seguida, dividir em uma Lista
def conv_ascii(string):
    string = str(text_to_bits(string))
    string = np.array(list(map(int, string)))    
    return string
```

```python
#Função para Converter os valores ASCII para String
def conv_string(string):
    
    x = ""
    for i in range(len(string)):
        x = x + str(string[i])
        
    n = int(x, 2)
    x = binascii.unhexlify('%x' % n)
    
    x = str(x)
    x = x.split("'")
    x = x[1]
    
    return(x)
```

<b> Exemplo de Recuperação de Conteúdo Hash (Dois Valores D-Hash)

```python
#HO = Hash Original e HA = Hash Alterado
ho = "7670795b33135a38"
ha = "7670795b33135a22"
```

```python
#Conversão dos valores Hash
ho = conv_ascii(ho)
ha = conv_ascii(ha)
```

```python
#Exibição dos Valores Pós-Conversão
print(ho)
print(ha)
```

```python
#Aplicação de Hopfield para Recuperação do Hash Differencial(D-Hash)
network = HopfieldNetwork(128)
network.train(ho)
```

```python
#Recuperação da Informação com Hopfield
hr = network.recall(ha);
```

<b> Conversão do Binário Recuperado Por Hopfield para String, além dos Binários Original e Alterado

```python
ho = str(conv_string(ho))
ha = str(conv_string(ha))
hr = str(conv_string(hr))
```

```python
#Resultados Finais:
print(ho + " - Hash Original Value")
print(ha + " - Hash Alternative Value")
```

```python
#Resultado do Valor Hash Recuperado:
print(hr + " - Hash Retrieved Value")
```

<b> Validação do Hash Recuperado pelo Cálculo da Distância de Hamming

```python
#Criando uma função para Calcular a Distância de Hamming entre os Valores Hash:
def hamming(hash1,hash2):
    L = len(hash1)
    distance = 0
    
    for i in range(L):
        if hash1[i] != hash2[i]:
            distance += 1
    
    print("Hamming Distance is: " + str(distance))
```

```python
#Calculando a distancia de Hamming entre os valores Hash:
hamming(ho,ha)
hamming(ho,hr)
```
