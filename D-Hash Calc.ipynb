```python
#Import das Bibliotecas
#Download da Lib D-Hash https://pypi.org/project/ImageHash/

import cv2
import imagehash
from PIL import Image
from matplotlib import pyplot as plot
```

```python
#Database Name Definition
db_name = "PalaceDatabase"
```

```python
#Path Definition
path = 'Image Database/' + db_name
```

```python
#Full Path Definition
fullpath = 'D:/Google Drive/Github/HNN-to-Content-Based-Hash-Retrieval/' + path
```

```python
#Image Extension Definition
ext = ".jpg"
```

```python
#Image Name Definition
image_name = str(input("Image Name: ") + ext)
```

```python
#Full Name Definition
full_image_name = str(fullpath + "/" + image_name)
```

```python
#Import and Plot of Image to Validation
img = cv2.imread(full_image_name)
img = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

plot.imshow(img,cmap='gray')
plot.show()
```

```python
#Path name Definition
name = path + "/" + image_name
```

```python
#Difference Hash (D-Hash) Calc
dhash = str(imagehash.dhash(Image.open(name)))
```

```python
print("D-Hash Value: " + str(dhash))
```
