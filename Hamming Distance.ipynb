```python
#Hamming Distance Function
def hamming(hash1,hash2):
    L = len(hash1)
    distance = 0
    
    for i in range(L):
        if hash1[i] != hash2[i]:
            distance += 1

    return(str(distance))
```

```python
#Average Hashing
hash1 = str(imagehash.average_hash(Image.open('LennaDatabase/original.jpg')))
ahash2 = str(imagehash.average_hash(Image.open('LennaDatabase/lenna2.jpg')))
ahash_dist = hamming(ahash1,ahash2)
```
