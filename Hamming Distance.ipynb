```python
#Import das Bibliotecas
#Download da Lib D-Hash https://pypi.org/project/ImageHash/

import cv2
import imagehash
from PIL import Image
from matplotlib import pyplot as plot
```

```python
#Database Name Definition
db_name = "ParkDatabase"
```

```python
#Path Definition
path = 'Image Database/' + db_name
```

```python
#Full Path Definition
fullpath = 'D:/Google Drive/Github/HNN-to-Content-Based-Hash-Retrieval/' + path
```

```python
#Image Extension Definition
ext = ".png"
```

```python
#Image Name Definition
image_original = str("Park" + ext)
image_altered = str(input("Altered Image Name: ") + ext)
```

```python
#Paths name Definition
name1 = path + "/" + image_original
name2 = path + "/" + image_altered
```

```python
#Dhash Calc
dhash1 = str(imagehash.dhash(Image.open(name1)))
dhash2 = str(imagehash.dhash(Image.open(name2)))
```

```python
#Results
print(str(dhash1) + " - Original Image")
print(str(dhash2) + " - Altered Image")
```

```python
#Hamming Distance Function
def hamming(hash1,hash2):
    L = len(hash1)
    distance = 0
    
    for i in range(L):
        if hash1[i] != hash2[i]:
            distance += 1

    return(str(distance))
```

```python
#Hamming Distance
hash_dist = hamming(dhash1,dhash2)
print("The Hamming Distance is: " + str(hash_dist))
```
